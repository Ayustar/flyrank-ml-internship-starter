## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

url = "https://raw.githubusercontent.com/Ayustar/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Filter to pages with real impressions and enough age
df_clean = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)].copy()

# Define the proxy target
df_clean['is_declining_label'] = (df_clean['trend_direction'] == 'down').astype(int)

# Show unit of analysis — one row = one content page
print(f"Total pages (unit of analysis): {len(df_clean)}")
print(f"\nDecline label distribution:")
print(df_clean['is_declining_label'].value_counts())
print(f"\nSample — what one row looks like:")
df_clean[['content_id', 'impressions_90d', 'content_age_days',
          'avg_position', 'ctr', 'trend_direction',
          'is_declining_label']].head(5)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can only check one or two conditions at a time —
for example: "flag pages older than 180 days with impressions > 500."
This ignores the interaction between age, position, CTR, engagement,
and trend direction simultaneously.

ML can learn which combination of signals actually predicts decline
across thousands of pages. The starter pipeline proved this directly:
- Hand-written baseline rule: Precision@50 = 0.24
- Random Forest model: Precision@50 = 0.74

That 3x lift came from the model finding signal combinations a fixed
rule cannot express. The output — a ranked list with scores — gives
a content team a prioritized queue they can act on immediately.
